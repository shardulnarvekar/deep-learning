```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
z=np.linspace(-10,10,100)
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))



fz_sigmoid=sigmoid(z)

def dsigmoid(z):
    return sigmoid(z)*(1-sigmoid(z))

dfz_sigmoid=dsigmoid(z)
```

```python
plt.figure(figsize=(12,4))
plt.plot(z,fz_sigmoid)
plt.plot(z,dfz_sigmoid)
plt.title("Sigmoid Graph")
plt.xlabel("z")
plt.ylabel("f(z)")
plt.grid()
plt.show()
```

```python
def tanh(z):
    return (np.exp(z)-np.exp(-z))/(np.exp(z)+np.exp(-z))

fz_tanh=tanh(z)

def dtanh(z):
    return 1-(tanh(z)*tanh(z))

dfz_tanh=dtanh(z)
```

```python
plt.figure(figsize=(12,4))
plt.plot(z,fz_tanh)
plt.plot(z,dfz_tanh)
plt.title("Tanh Graph")
plt.xlabel("z")
plt.ylabel("f(z)")
plt.grid()
plt.show()
```

```python
def relu(z):
    return np.maximum(0, z)

fz_relu=relu(z)

def drelu(z):
    for i in range (len(z)):
        if(z[i]>0):
            return 1
        else:
            return 0

dfz_relu=drelu(z)
```

```python
plt.figure(figsize=(12,4))
plt.plot(z,fz_relu)
plt.plot(z,dfz_relu)
plt.title("Relu Graph")
plt.xlabel("z")
plt.ylabel("f(z)")
plt.grid()
plt.show()
```

```python
def lrelu(z):
    alpha=0.1
    return np.maximum(alpha*z, z)

fz_lrelu=lrelu(z)
```

```python
plt.figure(figsize=(12,4))
plt.plot(z,fz_lrelu)
plt.title("Leaky Relu Graph")
plt.xlabel("z")
plt.ylabel("f(z)")
plt.grid()
plt.show()
```
