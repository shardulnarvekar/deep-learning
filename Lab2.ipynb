```python
input_data=np.array([[0,0],[0,1],[1,0],[1,1]])
```

tensor flow library is built on keras platform, is helps in defining the model, all the things are writien here in keras, we are going to use a sequintial model. For now, just understand that sequential model is forward propogating 

now we have to define all 3 layers, input, hidden1, hidden2,...... output.

here we are making a perceptron for XOR gate.

```python
MLP = tf.keras.Sequential( #this will initiate a blank model, we dont know, how many layers have how many perceptrons etc.
[
    #InputLayer
    tf.keras.Input(shape=(2,)),#This is one way of defining the input layer, defining that there are 2 neurons being created in the input layer.
    tf.keras.layers.Dense(3,activation='relu'),
    # tf.keras.Dense(2,activation='sigmoid') if we want to make another hidden layer of 2 neurons
    # similarly output has 1 neuron
    # so the last line will also be the same
    tf.keras.layers.Dense(1,activation='sigmoid')
    
]
)
#MLP stands for Multi Layered Perceptron
```

```python
MLP.compile(
    loss='binary_crossentropy',
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1), #here the learning rate is being used with the adam optimizer
    # now what matrics we are going to use to get the accuracy of the model
    # like accuracy, roc, auc, f1 score, recall, precision - all this for classification
    # regression we calculate r2 score
    metrics=['accuracy']
)
```

```python
# now we gonna to train the model

MLP.fit(
    input_data,
    output_data,
    epochs=100
)

# no need of splitting, train and test, because the data is small
```

```python
# dont again run the above code, the model will get better.
```

```python
ycap = MLP.predict(input_data)
#storing the predicted value
```

```python
ycap
```

notice, we have used sigmoid in the output layers, thus we are getting probabilities, now we keep a threshold, greater than 0.5 is 1 and less than 0.5 is 0

```python
for i in ycap:
    if i > 0.5:
        print(1)
    else:
        print(0)
```
