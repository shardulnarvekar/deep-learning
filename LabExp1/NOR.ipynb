```python
epoch=10
```

```python
#Learning Rate (eta)
```

```python
eta=0.1
```

```python
#now we define the activation function
```

```python
def activation(z):
    if z>0:
        return 1
    else:
        return 0
```

```python
len(input_data)
```

```python
for i in range (epoch):
    print(f'Epoch:{i}')
    for j in range(len(input_data)):
        print(input_data[j])
```

```python
for i in range (epoch):
    print(f'Epoch:{i}')
    print()
    for j in range(len(input_data)):
        x=input_data[j]
        yactual=output_data[j]
        z=np.dot(x,weights)+bias
        ypred = activation(z)
        error=yactual-ypred
        # print(error)
        #there is an error in the for iteration, thus we update the weight
        #refer the notebook
        weights=weights+(eta*error*x)
        bias=bias+eta*error
        #above is the formula to update the bias
        #note: the weights are getting updated after the 1st Epoch

        print("Actual Value: ",yactual)
        print("Predicted Value: ",ypred)
        print("Error: ",error)
        print()
    print()
    print()
```
